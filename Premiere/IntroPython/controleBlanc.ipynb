Les cases sont donc A1, A2, A3, ..., H7, H8.

Proposer un code qui écrit toutes les cases possibles.

### Exercice 2
* 1. Vous disposez d’une liste de nombres [2, 5, 7, 10, 13, 18].
* 2. Écrivez un programme en Python qui calcule la somme de tous les nombres pairs de la liste à l’aide d’une boucle for.
* 3. Affichez ensuite cette somme à l’écran.

<div id="editor"></div>

::: {.callout-tip collapse="true"}
## Correction
```python
liste_nombres = [2 , 5 , 7 , 10 , 13 , 18]
somme_nombres_pairs = 0
for nombre in liste_nombres :
    if nombre % 2 == 0: # On vérifie si le nombre est pair
        somme_nombres_pairs += nombre # On ajoute le nombre à la somme

print("La somme des nombres pairs de la liste est :", somme_nombres_pairs )
```

### Exercice 3
Écrire un programme en Python qui doit compter le nombre d’occurrences d’une lettre particulière dans une chaîne de caractères donnée. Par exemple, pour la chaîne de caractères 
"Bonjour tout le monde", il doit être retourné la valeur 3
pour la chaîne de caractères "o".

<div id="editor"></div>

::: {.callout-tip collapse="true"}
## Correction
```python
# Définition de la chaîne de caractères à traiter
chaine = " Bonjour tout le monde "
# Définition de la lettre que l’on cherche
lettre = ’o’
# Initialisation du compteur d’ occurrences de la lettre
nb_occurrences = 0
# Parcours de chaque caractère de la chaîne
for caractere in chaine :
    # Vérification si le caractère est égal à la lettre que l ’on cherche
    if caractere == lettre :
        # Si oui , on incrémente le compteur d’ occurrences
        nb_occurrences += 1
# Affichage du nombre d’ occurrences de la lettre
print(f"La lettre ’{ lettre }’ apparaît { nb_occurrences } fois dans la chaîne ’{ chaine } ’.")
```

### Exercice 5
Écrire un programme Python qui crée une table de multiplication à partir des nombres de 1 à 10. On doit créer une liste
(table) de listes(ligne). L’écriture de chaque liste de table doit donner ce résultat :

<pre id="editor">
</pre>

::: {.callout-tip collapse="true"}
## Correction
```python
table = []
for i in range(1 , 11) :
    ligne = []
    for j in range(1 , 11) :
        resultat = i * j
        ligne.append(resultat)
    table.append(ligne)
for ligne in table :
    print( ligne )
```

### Exercice 5
Écrire un programme en Python qui affiche une Pyramide complète avec des *, en fonction d'un entier N saisi par l'utilisateur.

Si l’on ajoute le paramètre end avec la valeur "", la fonction print() ne revient pas à la ligne.

Exemple : print("a", end="")

On peut multiplier un texte (une chaîne de caractères) pour le répéter plusieurs fois.

Par exemple : "a" * 10 donnera "aaaaaaaaaa"

Exemple de sortie:
```
Entrer le nombre de lignes: 4
   *
  ***
 *****
*******
```

<pre id="editor">
</pre>

::: {.callout-tip collapse="true"}
## Correction
Solution 1
```python
n = 4
for i in range(n):
    print(" "*((n-1)-i),"*"*((i*2)+1))

```
Solution 2
```python
n = 4
for i in range(n):
    for _ in range((n-1)-i):
        print(" ",end="")
    for _ in range(i*2):
        print("*",end="")
    print("*")

```
