```python
if condition :
    instruction 1
    istruction 2
elif condition 2:
    instruction 3
elif condition 3:
    instruction 4
else:
    instruction 5
```

## les boucles for
les boucle for sont un moyen de parcourire un enumerable , un enumerable va désigner les objets que l'on peut énumérer, c'est-à-dire les décomposer en une succession ordonnée d'éléments. On les appelle les énumérables ou les itérables (Python utilise le mot anglais iterable).

* la variable "NSI" (qui est de type String) est énumérable : on peut la décomposer en N, S, I.
* la variable [4, 3, 17] (qui est de type List) est énumérable : on peut la décomposer en 4, 3, 17.
* la variable 5 (qui est de type Int) n'est PAS énumérable : on ne peut pas la décomposer.

syntax :
```python
for elm in enumerable:
    instruction1 
    instruction 2
```
dans cette boucle les instruction indentée (qui on un espace a gauche)seron repetée pour chaque element de lenumerable 
exemple :
```python
for lettre in "NSI":
    print(lettre)
```
dans cette exemple la premiere iteration de la boucle lettre va prendre la valaur "N" ensuite il va executée print(lettre)
ensuiite vu quil ny a plus dinstruction va recommencer et lettre va prendre la valeur "S" et inci de suite jusqua quil ny ai plus de lettre dans lenumerable 

La fonction range
la fonction range va renvoyer un enumerable de nombre qui suivra une certaine logique grace au parametre que lon donnera

syntaxe :
```python
range(debut,fin,pas)
```
le code si dessu renvera un enumerable qui commencera par debut et qui avancera de pas en pas jusqua fin exclu
 on peux ommetre pas et par defaut il prendra la valeur 1
 on peux ommetre debut qui prendra par defaut la valeur 0

 ::: {.callout-note}
il est recomender de nommer la variable qui evolu au cours des iteration _ si on ne lutilise pas

## fonction 
les fonction serve a cree des bout de code reutilisable comme des outils que lon va declarée avan notre code que lon utilisera ensuite

syntaxe :
deffinition
```python
def Mafonction():
    print(1)
    print(2)
```

utilisation
```python
print(3)
Mafonction()
print(4)
Mafonction()
```

se qui donnera en sortie 
```
3
1
2
4
1
2
```

on peux ajoutée des parametre a une fontion

```python
def Mafonction(parametre1,parametre2):
    print(parametre1)
    print(parametre2)
```
utilisation
```python
print(3)
Mafonction(5,6)
print(4)
Mafonction(2,0)
```
se qui donnera en sortie 
```
3
5
6
4
2
0
```
enfin on peux ajoutée un return a la fonction qui permet de donnée une valeur a la fonction par defaut elle ne vaut rien

```python
def Mafonction(parametre1,parametre2):
    resultat = parametre1+parametre2
```
```python
print(Mafonction(1,2))
```
se qui donnera en sortie 
```
None
```
si on ajoute un return 
```python
def Mafonction(parametre1,parametre2):
    resultat = parametre1+parametre2
    return resultat
```
```python
print(Mafonction(1,2))
```
se qui donnera en sortie 
```
3
```
