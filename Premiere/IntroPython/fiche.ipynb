L'instruction conditionnelle if permet de soumettre l'exécution d'instructions à une condition donnée.
syntax :
```python
if condition :
    instruction 1
    istruction 2
```
il est possible dajoutée un else qui si la condition nest pas remplie(false) executera a la place le code
```python
if condition :
    instruction 1
    istruction 2
else :
    instruction 3
    istruction 4
```

enfin si lon veux teste plusieur condition a la suite on utilise elif  qui est june contraction de else et if se qui transforme un code 

```python
if condition :
    instruction 1
    istruction 2
else :
    if condition 2:
        if condition 3
            instruction 3
        else :
            istruction 4
```
 en 
```python
if condition :
    instruction 1
    istruction 2
elif condition 2:
    instruction 3
elif condition 3:
    instruction 4
else:
    instruction 5
```

## les boucles for
les boucle for sont un moyen de parcourire un enumerable , un enumerable va désigner les objets que l'on peut énumérer, c'est-à-dire les décomposer en une succession ordonnée d'éléments. On les appelle les énumérables ou les itérables (Python utilise le mot anglais iterable).

* la variable "NSI" (qui est de type String) est énumérable : on peut la décomposer en N, S, I.
* la variable [4, 3, 17] (qui est de type List) est énumérable : on peut la décomposer en 4, 3, 17.
* la variable 5 (qui est de type Int) n'est PAS énumérable : on ne peut pas la décomposer.
