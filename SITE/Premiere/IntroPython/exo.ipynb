### Exercice 2 : Nombres pairs
- Dans une fonction nommée Exo2A prenant en paramètre un entier naturel n, afficher les nombres pairs entre 2 et n inclus à l’aide uniquement d’une boucle bornée (for).
- Dans une fonction nommée Exo2B prenant en paramètre un entier naturel n, afficher les nombres pairs entre 2 et n inclus à l’aide uniquement d’une boucle non-bornée (while).

<div id="editor"></div>

::: {.callout-tip collapse="true"}
## Correction
```python
def Exo2A(n):
    for nbPaire in range(0,n,2):
        print(nbPaire)

def Exo2B(n):
    nbPaire = 0
    while nbPaire < n :
        print(nbPaire)
        nbPaire = nbPaire + 2
```

### Exercice 3 : Somme
Dans une fonction nommée Exo3 prenant en paramètre un entier naturel n, vérifier que n > 0 (précondition sur l’argument) puis renvoyer le résultat 0 + 1 + 2 + … + (n – 1) + n si la condition est vérifiée, et -1 sinon. 

<div id="editor"></div>

::: {.callout-tip collapse="true"}
## Correction
```python
def Exo3(n):
    if n < 0 :
        return -1
    somme = 0
    for i in range(n):
        somme = somme + i
    return somme
```

### Exercice 4 : Quel est mon âge ?
On représente une date par 3 nombres : j le jour du mois, m le numéro du mois (de 1 à 12) et a l’année. On considère ici, pour simplifier, que tous les mois comportent 30 jours (soit 360 jours par année).

- Ecrire une fonction Exo4A qui prend deux dates (j1, m1, a1 et j2, m2, a2) en arguments (dates après J.-C.) et renvoie True si la première date est antérieure (strictement avant) la seconde et False sinon.
Les préconditions sur les arguments doivent être vérifiées dans la fonction. Dans le cas contraire, la fonction ne renvoie rien.

- Ecrire une fonction Exo4B qui prend en arguments la date de naissance d’une personne et une autre date (la date de calcul) et renvoie l’âge de la personne à la date de calcul, en année (nombre à virgule). 
Les préconditions sur les arguments doivent être vérifiées dans la fonction. Dans le cas contraire, la fonction ne renvoie rien.

Exemple : Exo4B (24,3,1981,10,2,1996) renvoie environ 14,9 ans

<div id="editor"></div>

::: {.callout-tip collapse="true"}
## Correction
```python
def Exo4A(j1, m1, a1, j2, m2, a2):
    if not (1 <= j1 <= 30 and 1 <= m1 <= 12 and a1 >= 1):
        return

    if not (1 <= j2 <= 30 and 1 <= m2 <= 12 and a2 >= 1):
        return

    if a1 < a2:
        return True
    elif a1 == a2 and m1 < m2:
        return True
    elif a1 == a2 and m1 == m2 and j1 < j2:
        return True
    else:
        return False


def Exo4B(j1, m1, a1, j2, m2, a2):
    if not (1 <= j1 <= 30 and 1 <= m1 <= 12 and a1 >= 1):
        return

    if not (1 <= j2 <= 30 and 1 <= m2 <= 12 and a2 >= 1):
        return

    if not Exo4A(j1, m1, a1, j2, m2, a2):
        return

    naissance = a1 * 360 + m1 * 30 + j1
    calcul = a2 * 360 + m2 * 30 + j2

    age = (calcul - naissance) / 360

    return age

```

### Exercice 5 : Lancé de dé
On peut générer des entiers pseudo-aléatoires compris entre A et B inclus en python au moyen de la fonction randint(A,B), que l’on importe depuis la bibliothèque random.

- Dans une fonction sans argument nommée Exo5A, simuler 10 lancers d'un dé à 6 faces et renvoyer la moyenne des résultats obtenus.
- Dans une fonction sans argument nommée Exo5B appelant la fonction précédente, simuler 20 expériences dont chacune consiste à lancer 10 fois le dé, afficher la moyenne à chaque fois ainsi que la plus grande moyenne obtenue lors des 20 expériences à la fin.


<div id="editor"></div>

::: {.callout-tip collapse="true"}
## Correction
```python
import random

def Exo5A():
    somme = 0 
    for _ in range(10):
        somme = somme + random.randint(1,6)
    return somme/10

def Exo5B():
    somme = 0 
    for _ in range(20):
        somme = somme + Exo5A()
    return somme / 20
```
