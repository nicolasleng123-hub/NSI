## Condition
Une condition est une expression qui se termine par True (vrai) ou False (faux).

Liste des comparaisons

| Instr. | Utilité |  Instr. | Utilité | 
|-|-|-|-|
| > | supérieur | >= | supérieur ou égal |
| < | inférieur | <= | inférieur ou égal |
| == | égalité | != | différent |

Il existe aussi des opérateurs logiques qui s'appliquent sur des valeurs vraies ou fausses :

| Instr. | Utilité | 
|-|-|
| and | vrai si les deux opérandes sont vrais |
| or  | vrai si au moins un des opérandes est vrai |

L'instruction conditionnelle `if` permet d'exécuter du code seulement si une condition est vraie.

Syntaxe :
```python
if condition:
    instruction1
    instruction2
```

On peut ajouter un `else` qui s'exécute lorsque la condition est fausse :
```python
if condition:
    instruction1
    instruction2
else:
    instruction3
    instruction4
```

Pour tester plusieurs conditions successives on utilise `elif` (contraction de else + if) :
```python
if condition:
    instruction1
    instruction2
elif condition2:
    instruction3
elif condition3:
    instruction4
else:
    instruction5
```

## Les boucles `for`
Les boucles `for` permettent de parcourir un itérable (chaîne, liste, tuple, etc.).

Exemples :

* La chaîne "NSI" est itérable : on peut la décomposer en 'N', 'S', 'I'.
* La liste [4, 3, 17] est itérable : on peut la décomposer en 4, 3, 17.
* L'entier 5 n'est pas itérable.

Syntaxe :
```python
for elm in iterable:
    instruction1
    instruction2
```

Exemple :
```python
for lettre in "NSI":
    print(lettre)
```

La première itération affichera 'N',puis à la deuxième 'S', puis 'I'.

### La fonction `range`

`range` renvoie un itérable de nombres selon les paramètres donnés :

Syntaxe :
```python
range(debut, fin, pas)
```

La séquence commence à `debut`, avance de `pas` et s'arrête avant `fin` (fin exclue). `pas` vaut 1 par défaut et `debut` vaut 0 par défaut.

Note : si la variable d'itération n'est pas utilisée, il est recommandé de la nommer `_` :
```python
for _ in range(5):
    print("bonjour")
```

## La boucle `while`
La boucle `while` exécute des instructions tant que la condition est vraie.

Syntaxe :
```python
while condition:
    instruction1
    instruction2
```

Attention : une erreur fréquente est d'écrire une condition qui est toujours vraie ou toujours fausse, provoquant une boucle infinie ou jamais exécutée.

## Fonction
Une fonction est un bloc de code réutilisable que l'on définit avant de l'utiliser.

Définition :
```python
def ma_fonction():
    print(1)
    print(2)
```

Utilisation :
```python
print(3)
ma_fonction()
print(4)
ma_fonction()
```

Sortie :
```
3
1
2
4
1
2
```

On peut ajouter des paramètres :
```python
def ma_fonction(param1, param2):
    print(param1)
    print(param2)
```

```python
print(3)
ma_fonction(5, 6)
print(4)
ma_fonction(2, 0)
```

Sortie :
```
3
5
6
4
2
0
```

On peut aussi retourner une valeur avec `return`. Par défaut une fonction renvoie `None` si rien n'est retourné.

Exemple sans `return` :
```python
def ma_fonction(param1, param2):
    resultat = param1 + param2

print(ma_fonction(1, 2))
```

Affiche :
```
None
```

Avec `return` :
```python
def ma_fonction(param1, param2):
    resultat = param1 + param2
    return resultat

print(ma_fonction(1, 2))
```

Affiche :
```
3
```

Le mot-clé `return` arrête l'exécution de la fonction et renvoie la valeur fournie.

Une fonction peut accéder à une variable définie à l'extérieur (variable globale), mais le code extérieur ne peut pas accéder aux variables locales définies à l'intérieur de la fonction.
