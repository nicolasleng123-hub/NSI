## Le dictionnaire

Nous allons maintenant découvrir le dictionnaire. C’est un type construit très pratique, avec un accès rapide aux données grâce à des paires `clé : valeur`.

Un dictionnaire se déclare ainsi :
`{key1: value1, key2: value2, ...}`

On peut utiliser ce type pour associer un code binaire à un caractère.

Exemple :
```python
tableAscii = {
    "0100000": " ",
    "0100010": "!",
    "0100011": "%",
    "0100100": "'",
    "0100101": "#",
}
```

L’accès à une valeur se fait comme avec une liste, mais avec une clé :
`tableAscii["0100010"]` donne `!`.


````python
tableAscii = {
    f'{i:08b}': chr(i) for i in range(128)
}

assert tableAscii['01000001'] == 'A'
assert tableAscii['00110010'] == '2'
````

## UTF-8
### Principe de l’UTF-8

On commence par analyser le premier octet :

- si il commence par `0`, alors le caractère est encodé sur 1 octet : `01010010`
- sinon, le nombre de `1` avant le délimiteur `10` indique le nombre d’octets utilisés pour coder le caractère : `11110...`
- les octets suivants commencent tous par `10`
- enfin, on rassemble les bits après le délimiteur `10` pour reconstituer le caractère, puis on le décode grâce au dictionnaire donné

Exemple : `11110110101010111000101010101010`

- on sépare en octets : `11110110 10101011 10001010 10101010`
- on lit le premier octet : <span style="color:blue">111</span><span style="color:red">10</span>110 → 3 octets après celui-ci
- on obtient :  111<span style="color:red">10</span>110 <span style="color:red">10</span>101011 <span style="color:red">10</span>001010 <span style="color:red">10</span>101010 
- on fusionne les bits après le délimiteur `10` : `110101011001010101010`
- puis on le décode grâce au dictionnaire

Le cœur de l’UTF-8 consiste donc à repérer la longueur du caractère à partir du premier octet, puis à reconstruire ses bits pour retrouver le bon symbole.
L’UTF-8 est un codage en octets utilisé pour représenter presque tous les caractères du monde. Un octet est une suite de 8 bits.
